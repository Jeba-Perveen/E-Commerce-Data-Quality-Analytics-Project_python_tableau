#### Product Category Standardization

```python
df['product_category'].unique()
```

```python
df['product_name'].unique()
```

```python
df[['product_category','product_name']]
```

```python
def map_category(product_name):
    product_name = product_name.lower()
    if 'running shoes' in product_name:
        return 'Fashion'
    elif 'office chair' in product_name or 'coffee maker' in product_name:
        return 'Home'
    elif 'wireless mouse' in product_name or 'smartphone' in product_name:
        return 'Electronics'
    else:
        return 'Other'
df['category'] = df['product_name'].apply(map_category)
```

```python
df[['category','product_name']]
```

```python
# droping the product_category column as new category column is created with correct mapping
df.drop(columns = ['product_category'],inplace = True)
```

#### Handling inconsistent shipping days

```python
df['shipping_days'].unique()
```

```python
# Dropping the shipping_days because of inconsitent value
df.drop(columns = ['shipping_days'], inplace = True)
```

```python
# Recalculating the shipping days
df['shipping_days'] = (df['delivery_date']-df['order_date']).dt.days
```

```python
df.shape
```

```python
df[df['shipping_days']<0]
```

```python
swap_conditions = df['delivery_date']<df['order_date']
```

```python
df.loc[swap_conditions,['order_date','delivery_date']]= df.loc[swap_conditions,['delivery_date','order_date']].values
```

```python
# Replacing the negative values with nan
df.loc[df['shipping_days']<0,'shipping_days'] = np.nan
```
